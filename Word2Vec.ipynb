## Introduction to Word2Vec

Word2Vec is a word embedding technique used in NLP to convert words into numerical vectors.

It learns the meaning and relationships between words by analyzing the context in which they appear.

Words that have similar meanings or appear in similar contexts are represented by vectors that are close to each other.

### Two Main Architectures:
- CBOW (Continuous Bag of Words): Predicts a target word using its surrounding context words.
- Skip-gram: Predicts surrounding context words using a target word.

### Example:
Words like "King" and "Queen" have similar contexts, so their vectors are expected to be close in the vector space.

### Objective:
To understand how Word2Vec learns word relationships and represents words as numerical vectors.

## CBOW(Continuous Bag of Words)

CBOW is one of the two main architectures of Word2Vec used in Natural Language Processing (NLP).

It predicts a target word based on its surrounding context words. The model learns word relationships by understanding the words that appear together in a sentence.

### How CBOW Works:
1. Takes surrounding words as input.
2. Uses the context words to understand the sentence.
3. Predicts the missing or target word.

### Example:
Sentence: "I love learning NLP"

Context Words: "I", "love", "NLP"

Target Word: "learning"

### Key Idea:
CBOW uses surrounding words to predict the target word.

### Objective:
To understand how CBOW learns word representations from context.

```python
from gensim.models import Word2Vec

sentences = [
    ["king", "queen", "prince", "royal"],
    ["man", "woman", "boy", "girl"],
    ["cricket", "bat", "ball", "player"],
    ["chai", "coffee", "milk", "tea"]
]
```

```python
# CBOW Model
cbow_model = Word2Vec(
    sentences,
    vector_size=100,
    window=2,
    min_count=1,
    sg=0
)
```

```python
print("CBOW:", cbow_model.wv["king"][:5])
```

## Skip-gram Model

Skip-gram is another Word2Vec architecture used in Natural Language Processing (NLP).

Unlike CBOW, Skip-gram takes a target word as input and predicts its surrounding context words.

It learns relationships between words by identifying which words commonly appear near each other.

### How Skip-gram Works:
1. Takes a target word as input.
2. Identifies the surrounding context words.
3. Predicts the neighboring words.

### Example:
Sentence: "I love learning NLP"

Target Word: "learning"

Context Words: "I", "love", "NLP"

### Key Idea:
Skip-gram uses a target word to predict its surrounding context words.

### Objective:
To understand how Skip-gram learns word representations by predicting context words.

```python
# Skip-gram Model
skipgram_model = Word2Vec(
    sentences,
    vector_size=100,
    window=2,
    min_count=1,
    sg=1
)
```

```python
print("Skip-gram:", skipgram_model.wv["king"][:5])
```
