# CBOW vs Skip-gram: Two Architectures of Word2Vec

## Introduction to Word2Vec

Word2Vec is a popular Word Embedding technique used in Natural Language Processing (NLP).

It converts words into numerical vectors by learning relationships between words based on their surrounding context.

Word2Vec uses two main architectures:

1. **CBOW (Continuous Bag of Words):** Predicts a target word using its surrounding context words.
2. **Skip-gram:** Predicts surrounding context words using a target word.

Both architectures help computers learn meaningful word representations from text.

## CBOW(Continuous Bag of Words)

CBOW is one of the two main architectures of Word2Vec used in Natural Language Processing (NLP).

It predicts a target word based on its surrounding context words. The model learns word relationships by understanding the words that appear together in a sentence.

### How CBOW Works:
1. Takes surrounding words as input.
2. Uses the context words to understand the sentence.
3. Predicts the missing or target word.

### Example:
Sentence: "I love learning NLP"

Context Words: "I", "love", "NLP"

Target Word: "learning"

### Key Idea:
CBOW uses surrounding words to predict the target word.

### Objective:
To understand how CBOW learns word representations from context.

## Understanding Context Windows

A context window determines how many words before and after a target word are considered its context.

For example:

Sentence: "I love learning Natural Language Processing"

Target Word: "learning"

With a window size of 2:

- Left Context: "I", "love"
- Target Word: "learning"
- Right Context: "Natural", "Language"

The context window helps Word2Vec understand relationships between nearby words.

In Gensim, we define the window size using the `window` parameter.

```python
!pip install gensim
```

```python
sentence = ["I", "love", "learning", "Natural", "Language", "Processing"]

target_word = "learning"
window_size = 2

target_index = sentence.index(target_word)

left_context = sentence[max(0, target_index - window_size):target_index]
right_context = sentence[target_index + 1:target_index + window_size + 1]

print("Target Word:", target_word)
print("Context Words:", left_context + right_context)
```

## CBOW vs Skip-gram: Comparison

CBOW and Skip-gram learn word relationships in opposite directions.

| Feature | CBOW | Skip-gram |
|---|---|---|
| Input | Context words | Target word |
| Output | Target word | Context words |
| Learning Direction | Context → Target | Target → Context |
| Training Speed | Generally faster | Generally slower |
| Rare Words | Often works well with frequent words | Often useful for rare words |

### Key Takeaway

CBOW predicts a word from its surrounding context, whereas Skip-gram predicts surrounding words from a given target word.

These are general tendencies, not strict rules. Actual performance depends on the dataset and model settings.

```python
from gensim.models import Word2Vec

sentences = [
    ["king", "queen", "prince", "royal"],
    ["man", "woman", "boy", "girl"],
    ["cricket", "bat", "ball", "player"],
    ["chai", "coffee", "milk", "tea"]
]
```

```python
# CBOW Model
cbow_model = Word2Vec(
    sentences,
    vector_size=100,
    window=2,
    min_count=1,
    sg=0
)
```

```python
print("CBOW:", cbow_model.wv["king"][:5])
```

## Skip-gram Model

Skip-gram is another Word2Vec architecture used in Natural Language Processing (NLP).

Unlike CBOW, Skip-gram takes a target word as input and predicts its surrounding context words.

It learns relationships between words by identifying which words commonly appear near each other.

### How Skip-gram Works:
1. Takes a target word as input.
2. Identifies the surrounding context words.
3. Predicts the neighboring words.

### Example:
Sentence: "I love learning NLP"

Target Word: "learning"

Context Words: "I", "love", "NLP"

### Key Idea:
Skip-gram uses a target word to predict its surrounding context words.

### Objective:
To understand how Skip-gram learns word representations by predicting context words.

```python
# Skip-gram Model
skipgram_model = Word2Vec(
    sentences,
    vector_size=100,
    window=2,
    min_count=1,
    sg=1
)
```

```python
print("Skip-gram:", skipgram_model.wv["king"][:5])
```

```python
# Comparing CBOW and Skip-gram Embeddings

print("CBOW Vocabulary Size:", len(cbow_model.wv))
print("Skip-gram Vocabulary Size:", len(skipgram_model.wv))

print("\nCBOW Embedding Shape:", cbow_model.wv["king"].shape)
print("Skip-gram Embedding Shape:", skipgram_model.wv["king"].shape)

print("\nCBOW Word Vector (First 5 Values):")
print(cbow_model.wv["king"][:5])

print("\nSkip-gram Word Vector (First 5 Values):")
print(skipgram_model.wv["king"][:5])
```

## Conclusion

In this practical notebook, we explored the two main architectures of Word2Vec: CBOW and Skip-gram.

We learned that:

- CBOW predicts a target word using its surrounding context words.
- Skip-gram predicts surrounding context words using a target word.
- Context windows help identify relationships between nearby words.
- Both architectures generate numerical word embeddings.

Although CBOW and Skip-gram work in opposite directions, both help computers learn word relationships and represent words as numerical vectors.

**Key Takeaway:** CBOW learns from context to target, while Skip-gram learns from target to context.
